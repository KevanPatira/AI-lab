```python
arr = [[0,0,0],[0,0,0],[0,0,0]]

winA = False
winB = False

for i in range(9):
  while True:
    pos = int(input("Enter an empty cell no (0-8): "))
    if 0 <= pos <= 8 and arr[pos//3][pos%3] == 0:
      break
    print("Invalid position. Try again.")

    # Player A = 1
    # Player B = 2
  if i % 2 == 0:
    arr[pos//3][pos%3] = 1
  else:
    arr[pos//3][pos%3] = 2

  if i >= 4:
    # Check columns
    for j in range(3):
      if arr[0][j] != 0 and arr[0][j] == arr[1][j] == arr[2][j]:
        if arr[0][j] == 1:
          winA = True
        else:
          winB = True
        break
    # Check rows
    if not winA and not winB:
      for j in range(3):
        if arr[j][0] != 0 and arr[j][0] == arr[j][1] == arr[j][2]:
          if arr[j][0] == 1:
            winA = True
          else:
            winB = True
        break
    # Check diagonals
    if not winA and not winB:
      if arr[0][0] != 0 and arr[0][0] == arr[1][1] == arr[2][2]:
        if arr[0][0] == 1:
          winA = True
        else:
          winB = True

    elif arr[0][2] != 0 and arr[0][2] == arr[1][1] == arr[2][0]:
      if arr[0][2] == 1:
        winA = True
      else:
        winB = True

    # Stop the game if someone won
    if winA or winB:
      break
if winA:
    print("Player A wins")

elif winB:
    print("Player B wins")

else:
    print("Draw")
```
