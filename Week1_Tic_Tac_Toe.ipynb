```python
arr = [[0,0,0],[0,0,0],[0,0,0]]
winA = False
winB = False
for i in range(0, 9):
  while True:
    pos = int(input("Enter an empty cell no(0-8): "))
    if arr[pos//3][pos%3 - 1] == 0:
      break
  if i%2 ==0:
    arr[pos//3][pos%3 -1] = 1
  else:
    arr[pos//3][pos%3-1] = 2
  if i>=5:
    for j in range(0,3):
      if arr[0][j] == arr[1][j] & arr[0][j] == arr[2][j]:
        if arr[0][j] == 1:
          winA = True
          break
        else :
          winB = True
          break
      if arr[j][0] == arr[j][1] & arr[j][0] == arr[j][2]:
        if arr[j][0] == 1:
          winA = True
          break
        else :
          winB = True
          break
      if arr[0][0] == arr[1][1] == arr[2][2] or arr[0][2] == arr[1][1] == arr[2][0]:
        if arr[1][1] == 1:
          winA = True
          break
        else:
          winB = True
          break

if winA:
  print("Player A wins")
elif winB:
  print("Player B wins")
else:
  print("Draw")
```
