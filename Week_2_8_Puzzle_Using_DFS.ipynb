```python
def dfs(start, goal):
    stack = [(start, 0)]
    visited = set()
    max_branching = 0

    while stack:
        state, depth = stack.pop()

        if state == goal:
            return max_branching, depth

        if state in visited:
            continue

        visited.add(state)

        zero = state.index(0)
        row, col = zero // 3, zero % 3

        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
        children = []

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new = list(state)
                new[zero], new[r * 3 + c] = new[r * 3 + c], new[zero]
                children.append(tuple(new))

        max_branching = max(max_branching, len(children))

        for new_state in children:
            if new_state not in visited:
                stack.append((new_state, depth + 1))

    return None, None


start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

branching, depth = dfs(start, goal)

print("Branching Factor:", branching)
print("Depth:", depth)

```
