### 8-Puzzle Solver using A* Search (Simple Procedural Version)

This simple implementation uses:
1. **Plain tuples** to represent states (e.g., `(1, 2, 3, 0, 4, 5, 6, 7, 8)` where `0` is the empty space).
2. **Misplaced Tiles** as the heuristic function ($h(n)$).
3. **A* Search Algorithm** using a priority queue (via `heapq`) holding elements in the form: `(f_score, g_score, current_state, path_taken)`.

```python
import heapq

# Calculate Misplaced Tiles heuristic (excluding the blank space 0)
def misplaced_tiles(state, goal_state):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal_state[i]:
            count += 1
    return count

# Generate valid moves and resulting states
def get_neighbors(state):
    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)

    moves = []
    if row > 0: moves.append((-1, 0, 'Up'))
    if row < 2: moves.append((1, 0, 'Down'))
    if col > 0: moves.append((0, -1, 'Left'))
    if col < 2: moves.append((0, 1, 'Right'))

    for dr, dc, action in moves:
        new_row, new_col = row + dr, col + dc
        new_zero_idx = new_row * 3 + new_col

        # Swap empty tile (0) with the target tile
        new_state = list(state)
        new_state[zero_idx], new_state[new_zero_idx] = new_state[new_zero_idx], new_state[zero_idx]
        neighbors.append((tuple(new_state), action))

    return neighbors

# A* Search Algorithm without classes using Misplaced Tiles
def solve_8_puzzle(initial_state, goal_state):
    # Priority queue stores: (f_score, g_score, current_state, path)
    # path is a list of (action, state) tuples
    start_h = misplaced_tiles(initial_state, goal_state)
    open_list = [(start_h, 0, initial_state, [(None, initial_state)])]
    closed_set = set()

    while open_list:
        f, g, current_state, path = heapq.heappop(open_list)

        if current_state == goal_state:
            return path

        if current_state in closed_set:
            continue
        closed_set.add(current_state)

        for neighbor_state, action in get_neighbors(current_state):
            if neighbor_state in closed_set:
                continue
            new_g = g + 1
            new_h = misplaced_tiles(neighbor_state, goal_state)
            new_f = new_g + new_h
            new_path = path + [(action, neighbor_state)]
            heapq.heappush(open_list, (new_f, new_g, neighbor_state, new_path))

    return None

# Helper function to print board nicely
def print_board(state):
    for i in range(0, 9, 3):
        print(list(state[i:i+3]))
    print()
```

```python
# Define States using user input (converted to tuples so they are hashable)
initial_input = []
print("Initial State:")
for i in range(9):
  val = int(input(f"Enter value for tile {i+1}: "))
  initial_input.append(val)
initial_state = tuple(initial_input)

goal_input = []
print("Goal State:")
for i in range(9):
  val = int(input(f"Enter value for tile {i+1}: "))
  goal_input.append(val)
goal_state = tuple(goal_input)

print("\nInitial State:")
print_board(initial_state)

print("Goal State:")
print_board(goal_state)

solution = solve_8_puzzle(initial_state, goal_state)

if solution:
    print(f"Solved in {len(solution) - 1} moves:\n")
    for step, (action, state) in enumerate(solution):
        if action:
            print(f"Step {step}: Move {action}")
        else:
            print("Start State:")
        print_board(state)
    b = 3.17
    d = len(solution) - 1
    TC = b**d
    print(f"Branching Factor: {b}")
    print(f"Depth of Tree: {d}")
    print(f"Time Complexity: O({TC:.2f})")
    print(f"Space Complexity: O({TC:.2f})")
else:
    print("No solution found.")
```
